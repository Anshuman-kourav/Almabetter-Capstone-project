## Types of Data


```python
Types of Data:

    1.Qualitative Data(Categorical) --> Ordinal and Nominal
    2.Quantative Data(Numerical) --> Discrete and Continous

Discrete --> Finite Numbers of Data --> Whole Number
         --> Number of students in a class.
         --> Number of Bank Accounts

Continous Data --> Grouping is not possible --> Whole number + Float Number
               --> Price of a House
               --> Salary

Ordinal --> Categorial Data --> Order is important --> Give Rank
        --> good,bad,average
        --> often ,veryoften

Nominal --> Categorical Data --> Rank and Order is not imporant
        --> Gender --> Male and Female
        --> colors --> red,green,yellow
```

```python
--> Bar Plot is used to show the categorical data in form of bars.
```

```python
import matplotlib.pyplot as plt
X =['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday']
Y =[10,20,30,40,50,60,70]
plt.bar(X,Y) #barplot plot will be implemented
```

# Customization of Plots

```python
plt.title() --> used to provide title to our plot.
plt.xlabel() --> used to provide label on x axis
plt.ylabel() --> used to provide label on y axis
plt.show() --> used to remove extra interpretion of plot
plt.xlim() --> used to put our own limit in x axis
plt.ylim() --> used to put our own limit in y-axis
```

```python
x = [10,20,30,40,50,60]
y = [98,45,24,24,25,6]
plt.plot(x,y,color='red',linewidth=2,marker='o')
plt.xlim([0,150])
plt.ylim([0,150])
plt.xlabel('Distance(m/sec)')
plt.ylabel('Time(sec)')
plt.title('Time vs Speed Line Plot')
plt.show()
```

# Line Plot

```python
--> Line plot is used to make a series of data points.
--> Line plot is not effective with categorical data.
--> Line plot is very important with continous data.
--> Line plot is used to show trend over time
--> They are commonly used with shares,stock market,sales per year,weather
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
data1 = pd.read_csv('tips1.csv')
data1.head(5)
```

```python
sns.lineplot(data=data1,y='tip',x='total_bill',hue='smoker')
plt.title('Tip VS Total Bill')
plt.show()
```

```python
sns.lineplot(data=data1,y='smoker',x='sex')
plt.title('Tip VS Total Bill')
plt.show()
```
